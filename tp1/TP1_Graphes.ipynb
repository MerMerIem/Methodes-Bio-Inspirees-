## Partie 1: Bases de la théorie des graphes

```python
import networkx as nx
import numpy as np
```

### Question 1: Creation et ajout des sommets aux graphes G et DG

```python
# Creation des graphes
G = nx.Graph()
DG = nx.DiGraph()

# Ajout des sommets 1,2,3,4 et 5 a chaque sommet
G.add_nodes_from([i for i in range(1, 6)])
DG.add_nodes_from([i for i in range(1, 6)])
```

### Question 2: Affichages de sommets des graphes G et DG

```python
# Affichage des sommets des graphes G et DG
print("Sommets du graphe G : ", G.nodes())
print("Sommets du graphe DG : ", DG.nodes())
```

### Question 3: Suppression du sommet 1 du graphe G

```python
# Suppression du sommet 1 du graphe G
G.remove_node(1)
print("Sommets du graphe G après suppression : ", G.nodes())
print("Sommets du graphe DG : ", DG.nodes())
```

### Question 4: Ajout des arêtes au graphe G

```python
# Ajout des aretes entre les sommets du graphe G
G.add_edges_from([(2, 3), (2, 5), (3, 4), (4, 5)])

# Affichage des sommets du graphe G
print(G.edges)
```

### Question 5: Ajout des arcs au graphe DG

```python
# Ajout des aretes au graphe DG
DG.add_edges_from([(1, 3), (2, 3), (2, 4), (2, 5), (4, 5), (5, 1)])
print("Aretes du graphe DG : ", DG.edges())
```

### Question 6: Tracé des graphes G et DG

```python
import matplotlib.pyplot as plt

# Trace des graphes G et DG
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
nx.draw(G, with_labels=True, node_color="lightblue", node_size=700)
plt.title("Graphe G")

plt.subplot(1, 2, 2)
nx.draw(DG, with_labels=True, node_color="lightgreen", node_size=700, arrows=True)
plt.title("Graphe oriente DG")

plt.show()
```

### Question 7: Taille du graphe G

```python
# Taille de G (nombre d'aretes)
print("Taille de G : ", G.size())
```

### Question 8: Ordre des graphes G et DG

```python
# Ordre de G et DG (nombre de sommets)
print("Ordre de G : ", G.order())
print("Ordre de DG : ", DG.order())
```

### Question 9: Déterminer et afficher la matrice d’adjacence de G

```python
# Affichage de la matrice d'adjacence de G et DG
adj_matrix_G = nx.adjacency_matrix(G).todense()
print("Matrice d'adjacence de G :\n", adj_matrix_G)
```

### Question 10: Pondérer les arêtes de DG

```python
# Ponderation des aretes du graphe DG
for (u, v) in DG.edges():
    random = np.random.randint(1, 5)  # Génère un entier aléatoire entre 1 et 4
    DG.edges[u, v]['weight'] = random  # Assigne le poids à l'arête (u, v)

print("Aretes du graphe DG avec poids : ")
for (u, v, weight) in DG.edges(data='weight'):
    print(f"({u}, {v}) avec poids {weight}")
```

### Question 11: Déterminer et afficher la matrice d’adjacence de DG

```python
# Affichage de la matrice d'adjacence de DG
adj_matrix_DG = nx.adjacency_matrix(DG).todense()
print("Matrice d'adjacence de DG :\n", adj_matrix_DG)
```

### Question 12: Afficher les listes d’adjacence de DG

```python
# Affichage des listes d'adjacence de DG
adj_list_DG = {node: list(DG.neighbors(node)) for node in DG.nodes()}
print("Listes d'adjacence de DG :")
for node, neighbors in adj_list_DG.items():
    print(f"  {node}: {neighbors}")
```

## Partie 2: Parcours en profondeur

### Question 1: Implémentation de G en utilisant un dictionnaire


```python
# Chaque clé est un sommet, chaque valeur est la liste de ses voisins
# (lue ligne par ligne dans la matrice d'adjacence)
G = {
    'a': ['b', 'c'],
    'b': ['a', 'd', 'e'],
    'c': ['a', 'd'],
    'd': ['b', 'c', 'e'],
    'e': ['b', 'd', 'f', 'g'],
    'f': ['e', 'g'],
    'g': ['e', 'f', 'h'],
    'h': ['g']
}

# Affichage du dictionnaire
print("Graphe G (dictionnaire) :")
for sommet, voisins in G.items():
    print(sommet, "->", voisins)
```

```python
# Tracé du graphe pour vérifier visuellement
import networkx as nx
import matplotlib.pyplot as plt

graphe = nx.Graph(G)
pos = {'a': (0, 1), 'b': (1, 2), 'c': (1, 0), 'd': (2, 1),
       'e': (3, 2), 'g': (4, 1), 'h': (3, 0), 'f': (5, 2)}
nx.draw(graphe, pos, with_labels=True, node_color='lightblue',
        node_size=800, font_size=14, edge_color='gray')
plt.show()
```

### Question 2 : Fonction ParcourirProfondeur(Graphe, Sommet, Visitee)

```python
def ParcourirProfondeur(Graphe, Sommet, Visitee):
    # 1. On marque le sommet courant comme « visité »
    Visitee.append(Sommet)

    # 2. On parcourt ses voisins, un par un
    for voisin in Graphe[Sommet]:
        # 3. Si le voisin n'a pas encore été visité, on relance le parcours depuis lui
        if voisin not in Visitee:
            ParcourirProfondeur(Graphe, voisin, Visitee)

    # 4. Quand tous les voisins ont été traités, on renvoie la liste des sommets visités
    return Visitee
```

### Question 3 : Application de la fonction sur le graphe G


```python
#Parcours en profondeur depuis le sommet a
print('depuis a:')
visitee_a = ParcourirProfondeur(G, 'a', [])
print(visitee_a)

#parcours en profondeur depuis le sommet f
visitee_f = ParcourirProfondeur(G, 'f', [])
print('depuis f :')
print(visitee_f)
```
